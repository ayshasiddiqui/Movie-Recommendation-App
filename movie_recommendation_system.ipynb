```python
# ==========================================================
# Cell 23 - Import AST
# ==========================================================

import ast
```

```python
# ==========================================================
# Cell 24 - Convert Genres
# ==========================================================

def convert(text):
    L = []
    for i in ast.literal_eval(text):
        L.append(i['name'])
    return L
```

```python
# ==========================================================
# Cell 25 - Apply on Genres
# ==========================================================

movies['genres'] = movies['genres'].apply(convert)

movies.head()
```

```python
# ==========================================================
# Cell 26 - Apply on Keywords
# ==========================================================

movies['keywords'] = movies['keywords'].apply(convert)

movies.head()
```

```python
# ==========================================================
# Cell 27 - Convert Cast (Top 3 Actors)
# ==========================================================

def convert3(text):
    L = []
    counter = 0

    for i in ast.literal_eval(text):

        if counter != 3:
            L.append(i['name'])
            counter += 1
        else:
            break

    return L
```

```python
# ==========================================================
# Cell 28 - Apply on Cast
# ==========================================================

movies['cast'] = movies['cast'].apply(convert3)

movies.head()
```

```python
# ==========================================================
# Cell 29 - Extract Director
# ==========================================================

def fetch_director(text):

    L = []

    for i in ast.literal_eval(text):

        if i['job'] == 'Director':
            L.append(i['name'])
            break

    return L
```

```python
# ==========================================================
# Cell 30 - Apply on Crew
# ==========================================================

movies['crew'] = movies['crew'].apply(fetch_director)

movies.head()
```

```python
# ==========================================================
# Cell 31 - Convert Overview into List
# ==========================================================

movies['overview'] = movies['overview'].apply(lambda x: x.split())

movies.head()
```

```python
# ==========================================================
# Cell 32 - Remove Spaces from Genres
# ==========================================================

movies['genres'] = movies['genres'].apply(
    lambda x: [i.replace(" ", "") for i in x]
)
```

```python
# ==========================================================
# Cell 33 - Remove Spaces from Keywords
# ==========================================================

movies['keywords'] = movies['keywords'].apply(
    lambda x: [i.replace(" ", "") for i in x]
)
```

```python
# ==========================================================
# Cell 34 - Remove Spaces from Cast
# ==========================================================

movies['cast'] = movies['cast'].apply(
    lambda x: [i.replace(" ", "") for i in x]
)
```

```python
# ==========================================================
# Cell 35 - Remove Spaces from Crew
# ==========================================================

movies['crew'] = movies['crew'].apply(
    lambda x: [i.replace(" ", "") for i in x]
)
```

```python
# ==========================================================
# Cell 36 - Create Tags Column
# ==========================================================

movies['tags'] = movies['overview'] + \
                 movies['genres'] + \
                 movies['keywords'] + \
                 movies['cast'] + \
                 movies['crew']
```

```python
# ==========================================================
# Cell 37 - Create New DataFrame
# ==========================================================

new_df = movies[['movie_id', 'title', 'tags']]
```

```python
# ==========================================================
# Cell 38 - Display New Dataset
# ==========================================================

new_df.head()
```

```python
# ==========================================================
# Cell 39 - Convert List into String
# ==========================================================

new_df['tags'] = new_df['tags'].apply(lambda x: " ".join(x))
```

```python
# ==========================================================
# Cell 40 - Convert into Lowercase
# ==========================================================

new_df['tags'] = new_df['tags'].apply(lambda x: x.lower())
```

```python
# ==========================================================
# Cell 41 - Final Dataset
# ==========================================================

new_df.head()
```

```python
# ==========================================================
# Cell 42 - Final Shape
# ==========================================================

new_df.shape
```

```python
# ==========================================================
# Cell 43 - Import Required Libraries
# ==========================================================

from sklearn.feature_extraction.text import CountVectorizer
from nltk.stem.porter import PorterStemmer
from sklearn.metrics.pairwise import cosine_similarity
import pickle
```

```python
# ==========================================================
# Cell 44 - Create Porter Stemmer
# ==========================================================

ps = PorterStemmer()
```

```python
# ==========================================================
# Cell 45 - Create Stemming Function
# ==========================================================

def stem(text):
    y = []

    for i in text.split():
        y.append(ps.stem(i))

    return " ".join(y)
```

```python
# ==========================================================
# Cell 46 - Apply Stemming
# ==========================================================

new_df['tags'] = new_df['tags'].apply(stem)

new_df.head()
```

```python
# ==========================================================
# Cell 47 - Create CountVectorizer
# ==========================================================

cv = CountVectorizer(max_features=5000, stop_words='english')
```

```python
# ==========================================================
# Cell 48 - Convert Text into Vectors
# ==========================================================

vectors = cv.fit_transform(new_df['tags']).toarray()
```

```python
# ==========================================================
# Cell 49 - Check Vector Shape
# ==========================================================

vectors.shape
```

```python
# ==========================================================
# Cell 50 - Check Vocabulary
# ==========================================================

cv.get_feature_names_out()
```

```python
# ==========================================================
# Cell 51 - Calculate Cosine Similarity
# ==========================================================

similarity = cosine_similarity(vectors)
```

```python
# ==========================================================
# Cell 52 - Check Similarity Matrix Shape
# ==========================================================

similarity.shape
```

```python
# ==========================================================
# Cell 53 - Recommendation Function
# ==========================================================

def recommend(movie):

    movie_index = new_df[new_df['title'] == movie].index[0]

    distances = similarity[movie_index]

    movies_list = sorted(
        list(enumerate(distances)),
        reverse=True,
        key=lambda x: x[1]
    )[1:6]

    for i in movies_list:
        print(new_df.iloc[i[0]].title)
```

```python
# ==========================================================
# Cell 54 - Test Recommendation
# ==========================================================

recommend('Avatar')
```

```python
# ==========================================================
# Cell 55 - Another Test
# ==========================================================

recommend('Batman Begins')
```
